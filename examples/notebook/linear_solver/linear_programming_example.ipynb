
Linear optimization example.

```python
from ortools.linear_solver import pywraplp



def LinearProgrammingExample():
    """Linear programming sample."""
    # Instantiate a Glop solver, naming it LinearExample.
    solver = pywraplp.Solver.CreateSolver("GLOP")
    if not solver:
        return

    # Create the two variables and let them take on any non-negative value.
    x11 = solver.NumVar(0, solver.infinity(), "x11")
    x12 = solver.NumVar(0, solver.infinity(), "x12")
    x13 = solver.NumVar(0, solver.infinity(), "x13")
    x21 = solver.NumVar(0, solver.infinity(), "x21")
    x22 = solver.NumVar(0, solver.infinity(), "x22")
    x23 = solver.NumVar(0, solver.infinity(), "x23")

    print("Number of variables =", solver.NumVariables())

    # Constraint 0: 5 * (x11+x12+x13) + 5 * (2 * (x21+x22+x23)) <= 1500000
    solver.Add(5 * (x11+x12+x13) + 5 * (2 * (x21+x22+x23)) <= 1500000)

    # Constraint 1:2 * (x21+x22+x23) <= 200000
    solver.Add(2 * (x21+x22+x23) <= 200000)

    # Constraint 2: x11 + x21 <= 50000
    solver.Add(x11 + x21 <= 50000)

    # Constraint 3: x12 + x22 <= 30000
    solver.Add(x12 + x22 <= 30000)

    # Constraint 2: x13 + x23 <= 40000
    solver.Add(x13 + x23 <= 40000)

    # Constraint 2:  82𝑥11 + 98𝑥21 >= 87(𝑥11 + 𝑥21)
    solver.Add( 82*x11 + 98*x21 >= 87*(x11 + x21))

    # Constraint 2:  82𝑥12 + 98𝑥22 >= 89(𝑥12 + 𝑥22)
    solver.Add( 82*x12 + 98*x22 >= 89*(x12 + x22))

     # Constraint 2:  82𝑥13 + 98𝑥23  >= 92(𝑥13 + 𝑥23)
    solver.Add( 82*x13 + 98*x23 >= 92*(x13 + x23) )

    print("Number of constraints =", solver.NumConstraints())

    # Objective function: 6.70(𝑥11 + 𝑥21) + 7.20(𝑥12 + 𝑥22) + 8.10(𝑥13 + 𝑥23
    solver.Maximize(6.70*(x11 + x21) + 7.20*(x12 + x22) + 8.10*(x13 + x23))

    # Solve the system.
    print(f"Solving with {solver.SolverVersion()}")
    status = solver.Solve()

    if status == pywraplp.Solver.OPTIMAL:
        print("Solution:")
        print(f"Objective value = {solver.Objective().Value():0.1f}")
        print(f"x11 = {x11.solution_value():0.1f}")
        print(f"x12 = {x12.solution_value():0.1f}")
        print(f"x13 = {x13.solution_value():0.1f}")
        print(f"x21 = {x21.solution_value():0.1f}")
        print(f"x22 = {x22.solution_value():0.1f}")
        print(f"x23 = {x23.solution_value():0.1f}")
    else:
        print("The problem does not have an optimal solution.")

    print("\nAdvanced usage:")
    print(f"Problem solved in {solver.wall_time():d} milliseconds")
    print(f"Problem solved in {solver.iterations():d} iterations")


LinearProgrammingExample()

```
